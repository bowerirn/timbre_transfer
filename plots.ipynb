```python
import torch
import numpy as np
import matplotlib.pyplot as plt
```

```python
mel_clar = "results/mel/mel_clar_ckpt_e1999.pth"
mel_strings = "results/mel/mel_clar_ckpt_e1999.pth"
mel_piano = "results/mel/mel_piano_ckpt_e1999.pth"
mel_vibes = "results/mel/mel_vibes_ckpt_e1999.pth"

ts_clar = "results/ts/ts_clar_ckpt_e499.pth"
ts_strings = "results/ts/ts_clar_ckpt_e499.pth"
ts_piano = "results/ts/ts_piano_ckpt_e499.pth"
ts_vibes = "results/ts/ts_vibes_ckpt_e499.pth"
```

```python
mel_clar = torch.load(mel_clar, map_location="cuda", weights_only=False)
mel_strings = torch.load(mel_strings, map_location="cuda", weights_only=False)
mel_piano = torch.load(mel_piano, map_location="cuda", weights_only=False)
mel_vibes = torch.load(mel_vibes, map_location="cuda", weights_only=False)

ts_clar = torch.load(ts_clar, map_location="cuda", weights_only=False)
ts_strings = torch.load(ts_strings, map_location="cuda", weights_only=False)
ts_piano = torch.load(ts_piano, map_location="cuda", weights_only=False)
ts_vibes = torch.load(ts_vibes, map_location="cuda", weights_only=False)
```

```python
mel_clar_fads     = mel_clar["fads"]
mel_clar_losses   = mel_clar["epoch_losses"]

mel_strings_fads  = mel_strings["fads"]
mel_strings_losses= mel_strings["epoch_losses"]

mel_piano_fads    = mel_piano["fads"]
mel_piano_losses  = mel_piano["epoch_losses"]

mel_vibes_fads    = mel_vibes["fads"]
mel_vibes_losses  = mel_vibes["epoch_losses"]

ts_clar_fads      = ts_clar["fads"]
ts_clar_losses    = ts_clar["epoch_losses"]

ts_strings_fads   = ts_strings["fads"]
ts_strings_losses = ts_strings["epoch_losses"]

ts_piano_fads     = ts_piano["fads"]
ts_piano_losses   = ts_piano["epoch_losses"]

ts_vibes_fads     = ts_vibes["fads"]
ts_vibes_losses   = ts_vibes["epoch_losses"]
```

```python
plt.figure()
plt.plot(mel_clar_fads,    label="Clarinet")
plt.plot(mel_strings_fads, label="Strings", alpha=0.5)
plt.plot(mel_piano_fads,   label="Piano")
plt.plot(mel_vibes_fads,   label="Vibes")
plt.xlabel("Epoch")
plt.ylabel("FAD")
plt.title("Mel FADs Through Training")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.savefig(f"mel_fad.png", dpi=200)
```

```python
plt.figure()
plt.plot(mel_clar_losses,    label="Clarinet")
plt.plot(mel_strings_losses, label="Strings", alpha=0.5)
plt.plot(mel_piano_losses,   label="Piano")
plt.plot(mel_vibes_losses,   label="Vibes")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title("Mel Loss per Epoch")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.savefig(f"mel_loss.png", dpi=200)
```

```python
plt.figure()
plt.plot(ts_clar_fads,    label="Clarinet")
plt.plot(ts_strings_fads, label="Strings", alpha=0.5)
plt.plot(ts_piano_fads,   label="Piano")
plt.plot(ts_vibes_fads,   label="Vibes")
plt.xlabel("Epoch")
plt.ylabel("FAD")
plt.title("TS FADs Through Training")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.savefig(f"ts_fad.png", dpi=200)
```

```python
plt.figure()
plt.plot(ts_clar_losses,    label="Clarinet")
plt.plot(ts_strings_losses, label="Strings", alpha=0.5)
plt.plot(ts_piano_losses,   label="Piano")
plt.plot(ts_vibes_losses,   label="Vibes")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title("TS Loss per Epoch")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.savefig(f"ts_loss.png", dpi=200)
```
